```python
# 8-Puzzle Problem using Depth-First Search (DFS)
print("1WA24CS064,ASHISH")
INITIAL_STATE = (
    (1, 2, 3),
    (0, 4, 6),
    (7, 5, 8)
)

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)


MOVES = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}

def find_zero(state):
    """Find row and column index of the blank space (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    """Generate all valid successor states from current state."""
    neighbors = []
    r, c = find_zero(state)

    for move_name, (dr, dc) in MOVES.items():
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            # Convert tuple of tuples to list of lists for swapping
            new_board = [list(row) for row in state]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            # Convert back to immutable tuple for hashing/visited set
            neighbors.append((tuple(tuple(row) for row in new_board), move_name))

    return neighbors

def dfs_8_puzzle(initial, goal, max_depth=20):
    """Depth-First Search with Depth Limit to avoid infinite recursion."""
    # Stack stores tuples of: (current_state, path_history)
    stack = [(initial, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path + [state]

        if state not in visited and len(path) <= max_depth:
            visited.add(state)

            # Explore valid move choices
            for neighbor, _ in get_neighbors(state):
                if neighbor not in visited:
                    stack.append((neighbor, path + [state]))

    return None

def print_state(state, step_num):
    print(f"Step {step_num}:")
    for row in state:
        print(" ".join(str(val) if val != 0 else "0" for val in row))
    print()

if __name__ == "__main__":
    solution_path = dfs_8_puzzle(INITIAL_STATE, GOAL_STATE)

    if solution_path:
        print(f"Goal Reached in {len(solution_path) - 1} steps!\n")
        for step, state in enumerate(solution_path):
            print_state(state, step)
    else:
        print("No solution found within the specified depth limit.")
```

```python
# 8-Puzzle Problem using IDDFS (Printing Step-by-Step for Each Depth Limit)
print("1WA24CS064,ASHISH")
INITIAL_STATE = (
    (1, 2, 3),
    (0, 4, 6),
    (7, 5, 8)
)

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

MOVES = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}

def find_zero(state):
    """Find row and column index of the blank space (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    """Generate all valid successor states from current state."""
    neighbors = []
    r, c = find_zero(state)

    for move_name, (dr, dc) in MOVES.items():
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [list(row) for row in state]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append(tuple(tuple(row) for row in new_board))

    return neighbors

def print_board(state):
    for row in state:
        print(" ".join(str(val) for val in row))
    print()

def depth_limited_search(state, goal, limit, path, visited):
    """Depth-Limited Search that prints every state evaluated at current depth."""
    path.append(state)

    current_depth = len(path) - 1
    indent = "  " * current_depth
    print(f"{indent}[Depth {current_depth}] Exploring state:")
    for row in state:
        print(f"{indent}  {' '.join(str(val) for val in row)}")
    print()

    if state == goal:
        return True

    if limit <= 0:
        path.pop()
        return False

    visited.add(state)

    for neighbor in get_neighbors(state):
        if neighbor not in visited:
            if depth_limited_search(neighbor, goal, limit - 1, path, visited):
                return True

    # Backtrack
    visited.remove(state)
    path.pop()
    return False

def iddfs_8_puzzle(initial, goal, max_limit=10):
    """Iterative Deepening DFS showing steps limit-by-limit."""
    for depth in range(max_limit):
        print("=" * 50)
        print(f"--- STARTING SEARCH WITH DEPTH LIMIT = {depth} ---")
        print("=" * 50)

        path = []
        visited = set()

        if depth_limited_search(initial, goal, depth, path, visited):
            print(f"\nSUCCESS! Goal found at depth limit {depth}.\n")
            return path

        print(f"Limit {depth} exhausted. Goal not found at this limit.\n")

    return None

# Driver Code
if __name__ == "__main__":
    solution_path = iddfs_8_puzzle(INITIAL_STATE, GOAL_STATE, max_limit=10)

    if solution_path:
        print("=" * 50)
        print("FINAL OPTIMAL SOLUTION PATH:")
        print("=" * 50)
        for step, state in enumerate(solution_path):
            print(f"Step {step}:")
            print_board(state)
    else:
        print("No solution found within max depth limit.")
```

```python
print("1WA24CS064,ASHISH")
import heapq

def misplaced_tiles(state, goal):
    count = 0
    for r in range(3):
        for c in range(3):
            if state[r][c] != 0 and state[r][c] != goal[r][c]:
                count += 1
    return count

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    neighbors = []
    r, c = find_blank(state)
    moves = [
        (0, -1, "Left"),
        (-1, 0, "Up"),
        (0, 1, "Right"),
        (1, 0, "Down")
    ]
    for dr, dc, move_name in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [list(row) for row in state]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append((tuple(tuple(row) for row in new_board), move_name))
    return neighbors

def print_matrix(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))

def solve_a_star(initial, goal):
    # Priority queue stores: (f, g, current_state, path_records)
    # path_records stores tuples of (state, move_name, h)
    h_init = misplaced_tiles(initial, goal)
    queue = [(0 + h_init, 0, initial, [(initial, "Initial", h_init)])]
    visited = set()

    while queue:
        f, g, current_state, path = heapq.heappop(queue)

        if current_state in visited:
            continue
        visited.add(current_state)

        if current_state == goal:
            return path

        for neighbor, move in get_neighbors(current_state):
            if neighbor not in visited:
                h = misplaced_tiles(neighbor, goal)
                next_f = (g + 1) + h
                queue.append((next_f, g + 1, neighbor, path + [(neighbor, move, h)]))
    return None

# --- Configuration matching the 2nd image ---
INITIAL_STATE = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

GOAL_STATE = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

solution = solve_a_star(INITIAL_STATE, GOAL_STATE)

if solution:
    for g, (state, move, h) in enumerate(solution):
        print(f"Move: {move}")
        print(f"g = {g}")
        print(f"h = {h}")
        print(f"f = {g + h}")
        print_matrix(state)
        print("-" * 20)
```
